**8. CREATING A FUNCTION**

```r
classify_grade <- function(grade) {
  if (grade >= 90) {
    classification <- "Excellent"
  } else if (grade >= 85) {
    classification <- "Very Good"
  } else if (grade >= 80) {
    classification <- "Good"
  } else if (grade >= 75) {
    classification <- "Passed"
  } else {
    classification <- "Failed"
  }
  return(classification)
}
```

```r
example_grade <- 75
result_classification <- classify_grade(example_grade)
cat("Grade:", example_grade, "-", result_classification, "\n")

example_grade_2 <- 90
result_classification_2 <- classify_grade(example_grade_2)
cat("Grade:", example_grade_2, "-", result_classification_2, "\n")
```

**11. CREATING A SIMPLE RULE-BASED SYSTEM**

```r
classify_risk <- function(grade_input, attendance_input) {
  if (attendance_input < 75 & grade_input < 75) {
    return("High Risk")
  } else if (attendance_input < 85 | grade_input < 80) {
    return("Moderate Risk")
  } else {
    return("Low Risk")
  }
}
```

**17. MINI KRR SYSTEM**

```r
# Pipeline: Input -> Data -> Rules -> Reasoning -> Classification -> Recommendation
# ==========================================

# Step 1: Input & Data Definition
student_id <- "S002"
attendance <- 72
grade <- 70

# Step 2: Reasoning System (Rules Engine Function)
krr_evaluate_student <- function(student_id, attendance, grade) {

  # Rule-Based Reasoning for Classification
  if (attendance < 75 & grade < 75) {
    risk <- "High Risk"
    recommendation <- "Immediate academic intervention"
  } else if (attendance < 85 | grade < 80) {
    risk <- "Moderate Risk"
    recommendation <- "Academic monitoring"
  } else if (attendance >= 85 & grade >= 80) {
    risk <- "Low Risk"
    recommendation <- "Continue regular monitoring"
  } else {
    risk <- "Unknown"
    recommendation <- "Check data validity"
  }

  # Answer Generation
  cat("==========================================\n")
  cat("          MINI KRR SYSTEM OUTPUT          \n")
  cat("==========================================\n")
  cat("Student ID    :", student_id, "\n")
  cat("Attendance    :", attendance, "%\n")
  cat("Grade         :", grade, "\n")
  cat("------------------------------------------\n")
  cat("Classification:", risk, "\n")
  cat("Recommendation:", recommendation, "\n")
  cat("==========================================\n")
}

# Step 3: Run the System to Answer the Question
krr_evaluate_student(student_id, attendance, grade)
```
